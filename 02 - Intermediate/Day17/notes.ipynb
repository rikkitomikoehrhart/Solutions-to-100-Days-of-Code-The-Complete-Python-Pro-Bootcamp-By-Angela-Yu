## Creating Classes

```python
class User:
    def __init__(self, user_id, username):
        self.id = user_id
        self.username = username
        self.followers = 0
        self.following = 0

    def follow(self, user):
        user.followers += 1
        self.following += 1

user_1 = User("001", "rikkitomikoehrhart")
print(user_1.username)
print(user_1.id)
print(user_1.followers)
```

- **Attributes** are variables attached to an object. 
- **Methods** are functions attached to an object.
- **Initialize** - to set (variables) to their starting values at the beginning of a program

```python
class Car:
    def __init__(self, seats):
        self.seats = seats

    def enter_race_mode(self):
        self.seats = 2

my_car = Car(5)
print(my_car.seats)
my_car.enter_race_mode()
print(my_car.seats)
```
