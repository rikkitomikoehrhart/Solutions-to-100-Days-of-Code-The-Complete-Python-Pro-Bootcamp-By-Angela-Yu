# Day 12 - Beginner - Scope and Number Guessing Game

````python
enemies = 1

def increase_enemies():
    enemies = 2
    print(f"enemies inside function {enemies}")

increase_enemies()
print(f"enemies outside function: {enemies}")
````

## Local Score
- what exists inside functions
- can be used by the function it is in
  
## Global Score
- exists outside 
- to be used globally

## No Block Scope in Python
- any variable created inside if, while, or for loop indented block is in it's local or global scope

## Global keyword
- `global` keyword 
- initiate the global scope variable:

````python
enemies = 1

def increase_enemies():
    global enemies
    enemies = 2
    print(f"enemies inside function {enemies}")

increase_enemies()
print(f"enemies outside function: {enemies}")
````

> NOTE: you don't want to be using the `global` keyword often because it can be confusing and there are more acceptable ways of changing variables from inside functions

## CONSTANTS
- variables that you never want to change
- naming convention is to put it in uppercase:
