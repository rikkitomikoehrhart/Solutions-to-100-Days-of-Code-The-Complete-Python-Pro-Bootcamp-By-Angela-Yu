## Mathematical Operations
- `+` - addition
- `-` - subtraction
- `*` - multiplication
- `/` - integer division
- `//` - float division
- `**` - exponent 

> NOTE: Python has Implicit Type Casting - when the Python interpreter automatically converts one data type to another without any manual intervention from the programmer

````python
print(123 + 456)
print(7 - 3)
print(3 * 2)
print(5 / 3)
print(5 // 3)
print(2 ** 3)
````

> NOTE: Python follows PEMDAS 
> Multiplication and Division are equally as important and it will run left to right
> Addition and Subtraction are equally as important and it will run left to right

````python
height = 1.65 
weight = 84

# Write your code here.
# Calculate the bmi using weight and height.
bmi = (weight) / (height * height)

print(bmi)
````

## Number Manipulation
- Flooring a number - placing a floating point number in `int()` which just removes the decimal and everything after
- Using `round()` rounds the number to the nearest whole number
  
example:

````python
print(bmi)
print(int(bmi))
print(round(bmi))
````

````python
# round to a certain decimal point, here the 2nd
print(round(bmi, 2))
````

### Shorthand
- `+=` add to itself
- `-=` subtract from itself
- `*=` multiply to itself
- `/=` and `//=` divide from itself

## F String

````python
print(f"Your BMI is: {round(bmi, 2)}")
````
