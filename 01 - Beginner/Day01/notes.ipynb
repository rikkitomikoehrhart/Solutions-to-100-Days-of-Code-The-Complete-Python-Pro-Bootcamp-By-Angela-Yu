## String Manipulation
- `\n` - new line character
- `+` - used to concatenate strings

````python
print("Hello World!\nHello World!\nHello World!")
print("Hello" + " " + "World!")
````

## Variables
- a way to story values to refer to throughout the code

````python
glass1 = "milk"
glass2 = "juice"

temp = glass1
glass1 = glass2
glass2 = temp
````

## Variable Naming
- names should be descriptive
- no spaces
- don't start with numbers
- don't use keywords
- if using multiple words, use _ between words
