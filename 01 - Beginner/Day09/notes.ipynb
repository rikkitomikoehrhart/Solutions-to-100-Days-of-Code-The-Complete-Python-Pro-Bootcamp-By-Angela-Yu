## Dictionaries
```python
dictionary = { 
    Key1: Value,
    Key2: Value,
    Key3: Value,
}

# Add to existing dictionary
dictionary[Key4] = value

# Reassign an existing key a new value
dictionary[Key4] = Value

# Loop through a dictionary
for key, value in dictionary:
    # do something with the key and/or value

# Empty dictionary / initiate an empty dictionary
dictionary = {} 
```



## Nesting
```python
nested_dictionary = {
    Key1: [ List ],
    Key2: { Dictionary },
    Key3: Value
}
```


````python
travel_log = {
    "France": ["Paris", "Lille", "Dijon"],
    "Germany": ["Stuttgart", "Berlin"]
}

print(travel_log["France"][1])
````

````python
nested_list = [ "A", "B", [ "C", "D" ]]

print(nested_list[2][1])
````

````python
travel_log = {
    "France": {
        "cities_visited": ["Paris", "Lille", "Dijon"],
        "total_visits": 12
    },
    "Germany": {
        "cities_visited": ["Stuttgart", "Hamburg", "Berlin"],
        "total_visits": 5
    },
}

print(travel_log["Germany"]["cities_visited"][2])
````
