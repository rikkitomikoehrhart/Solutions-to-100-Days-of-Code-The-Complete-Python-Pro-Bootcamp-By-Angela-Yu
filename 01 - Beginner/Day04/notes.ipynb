## Randomization
The one python uses is called Mersenne Twister, it is a pseudorandom number generator algorithm. 

````python
import random
````

````python
if random.randint(1, 10) % 2 == 0:
    print("Heads")
else:
    print("Tails")
````

## Python Lists
- variables hold single pieces of data
- lists store groups of data
- stored in square brackets
`fruits = ["apples", "grapes"]`

````python
states_of_america = ["Texas", "Tennessee", "California", "Washington", "New York"]

print(states_of_america[0])
````

```python
# Add to list
states_of_america.append("Georgia")
states_of_america.extend("New Jersey", "Louisiana")
```

````python
friends = ["Alice", "Bob", "Charlie", "David", "Emanuel"]

print(random.choice(friends))
````

## Nested Lists

````python
fruits = ["Strawberries", "Apples", "Oranges"]
veggies = ["Spinach", "Broccoli", "Cucumber"]

foods = [fruits, veggies]

print(foods)
````
