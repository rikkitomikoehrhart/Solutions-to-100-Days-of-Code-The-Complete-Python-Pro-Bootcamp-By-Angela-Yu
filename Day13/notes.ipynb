```python
def my_function():
    for i in range(1, 20):
        if i == 20:
            print("You got it!")
```

- The way `range` works is it starts at the first parameter and steps up to just before the second one.
- So the above code bug is that range never reaches 20

**FIX**: change `range` to `range(1, 21)`

```python
from random import randint
dice_imgs = ["❶", "❷", "❸", "❹", "❺", "❻"]

dice_num = randint(1, 6)
print(dice_imgs[dice_num])
```

- The way index works is that it starts at 0 (or the spot right before the item, then when it counts over it adds one, which is the starting of the second item, etc.)
- Since `randint` in the code is starting at 1 and ending at 6, there are two bugs:
  - ❶ will never be an answer
  - And there isn't an index 6 (which will produce an `IndexError`)
  
**FIX**: Change `randint` to `randint(0, 5)`

```python
year = 1994

if year > 1980 and year < 1994:
    print("You are a Millennial.")
elif year > 1994:
    print("You are a Gen Z.")
```

- This problem occurs because the Millennial statement only happens if the year is less than 1994 and the Gen Z happens if year is greater than 1994
- No block has 1994 as part of its condition

**FIX**: change the `>` or `<` to `>=` or `<=` depending on if you are adding 1994 to the Millennial block or the Gen Z block

## Try / Catch Block

```python
age = 0
try:
    age = int(input("How old are you? "))
except ValueError:
    print("You have typed in an invalid number. Please try again with a numerical value, as in 18.")

if age > 18:
    print(f"You can drive at age {age}")
```

```python
words_per_page = 0
pages = int(input("Number of pages: "))
words_per_page == int(input("Number of words per page: "))
total_words = pages * words_per_page

print(f"pages = {pages}")
print(f"words_per_page = {words_per_page}")
print(total_words)
```

- Use `print()` statements to follow data to figure out logical errors
