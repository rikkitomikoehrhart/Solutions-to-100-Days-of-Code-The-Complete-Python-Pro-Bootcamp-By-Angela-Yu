## Comparison Operators
| OPERATOR | MEANING |
| -------- | ------- |
| > | Greater Than |
| < | Less Than |
| >= | Greater Than or Equal To |
| <= | Less Than or Equal To |
| == | Equal To |
| != | Not Equal To | 



> NOTE a single = is an assignment operator but a double == is an equal check

## Modulo Operator

` 10 % 3 ` = 1
Modulo is the remainder operator.


````python
num = 13

if num % 2 == 0:
    print("Even Number")
else: 
    print("Odd Number")
````

## if / elif / else

```
if condition:
    do this
elif condition2:
    do this
...
...
...
else:
    do this

````python
weight = 85
height = 1.85

bmi = weight / (height ** 2)


if bmi < 18.5:
    print("underweight")
elif bmi < 24.9:
    print("normal weight")
else:
    print("overweight")
````

## Logical Operators

```python
A and B
C or D
not E
```

````python
age = 40

if age <= 12:
    print("You're a kid!")
elif age >= 13 and age <= 18:
    print("You're a teenager")
elif age >= 19 and age <= 30:
    print("You're a young adult!")
elif age >= 31 and age < 60:
    print("You are an adult")
else:
    print("You're old!")
````
