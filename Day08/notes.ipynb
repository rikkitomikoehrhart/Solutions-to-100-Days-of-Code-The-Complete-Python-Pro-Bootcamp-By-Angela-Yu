````python
def greet(name):
    print(f"Hello {name}")
    print("How do you do?")

greet("Rikki")
````

**Parameter** -- the local variable being made in the function scope



**Argument** -- the actual data being passed through

## Function with more than one parameter

````python
def greet_with(name, location):
    print(f"Hello, {name}!")
    print(f"What's it like in {location}?")

greet_with("Rikki", "Austin, TX")
````
