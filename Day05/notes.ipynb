## For Loop
```python
for item in list_of_items:
    #Do something to each item
```

````python
fruits = ["Apple", "Peach", "Pear"]
for fruit in fruits:
    print(fruit)
````

````python
student_scores = [150, 142, 185, 120, 171, 184, 149, 24, 59, 68, 199, 78, 65, 89]

sum = 0
for score in student_scores:
    sum += score

print(sum)

max_score = 0
for score in student_scores:
    if score > max_score:
        max_score = score

print(max_score)
````

## Range
```python
for number in range(a, b):
    print(number)
```

````python
for num in range(1, 101):
    if (num % 5 == 0) and (num % 3 == 0):
        print("FizzBuzz")
    elif (num % 5 == 0):
        print("Buzz")
    elif (num % 3 == 0):
        print("Fizz")
    else:
        print(num)
````
